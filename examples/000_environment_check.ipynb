# Checking your WS4 installation

This notebook checks that the pieces are where they should be: the interpreter, the
package, the command line, and the compiler. It writes only into `tmp/`.

## The interpreter and the package

```python
import subprocess


def run(command: str) -> None:
    """Run a WS4 command, printing what it printed."""
    result = subprocess.run(command, shell=True, text=True, capture_output=True)
    print((result.stdout or result.stderr).strip())
    result.check_returncode()
```

```python
import importlib.metadata
import sys

import ws4

print("python", sys.version.split()[0])
print("import ", ws4.__name__)
print("ws4    ", importlib.metadata.version("ws4"))
```

## The optional numerics

`numba` compiles the patch-state kernels; `scipy` is needed for `--mip-bound`.
Both are optional, and both are reported as found or absent.

```python
for module in ("numba", "scipy", "geopandas"):
    try:
        imported = __import__(module)
        print(f"{module:10} {getattr(imported, '__version__', 'present')}")
    except ImportError:
        print(f"{module:10} not installed (optional)")
```

## The command line

```python
run("ws4 --help")
```

## The compiler

`ws4 build` compiles the recipe and inventory in `examples/synthetic` into a dataset, and
`ws4 validate` checks it against its manifest and schema invariants.

```python
run("ws4 build --xml examples/synthetic/recipe.xml --fragments examples/synthetic/fragments.csv --out tmp/examples/compiled")
```

```python
run("ws4 validate --dataset tmp/examples/compiled")
```

If both commands print a destination and no error, the install is complete. Continue with
`010_quickstart.ipynb`.
